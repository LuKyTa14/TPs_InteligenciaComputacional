```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import KFold, cross_val_score
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.base import clone

# Datos
digits = load_digits()
X = digits.data
y = digits.target

# K=5
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Modelos
modelos = {
    "MLP (Red Neuronal)": MLPClassifier(hidden_layer_sizes=(100,), max_iter=1000, learning_rate_init=0.01, activation="logistic", random_state=42),
    "Naive Bayes": GaussianNB(),
    "LDA": LinearDiscriminantAnalysis(),
    "KNN (K=5)": KNeighborsClassifier(),
    "Árbol de Decisión": DecisionTreeClassifier(random_state=42),
    "SVM (Soporte Vectorial)": SVC(random_state=42),
}


print("Comparacion de Clasificadores (5-Fold)")
print("-" * 65)
print(f"{'Clasificador':<25} | {'Media de Acierto':<25} | {'Varianza':<25}")
print("-" * 65)

# Entrenamiento y Evaluación
for nombre, modelo in modelos.items():
    # Evaluacion automatica
    scores = cross_val_score(modelo, X, y, cv=kf)
    
    media = np.mean(scores) * 100
    varianza = np.var(scores)

    print(f"{nombre:<25} | {media:.2f} {'%':<19} | {varianza:<10.6f}")

print("-" * 65)
```

```python
print("Comparacion de Clasificadores (5-Fold)")
print("-" * 65)
print(f"{'Clasificador':<25} | {'Media de Acierto':<25} | {'Varianza':<25}")
print("-" * 65)

for nombre, modelo in modelos.items():
    aciertos = []
    for train_idx, test_idx in kf.split(X):
        # modelo = clone(modelo)
        modelo.fit(X[train_idx], y[train_idx])
        acierto = modelo.score(X[test_idx], y[test_idx])
        aciertos.append(acierto)
    
    aciertos = np.array(aciertos)
    media = np.mean(aciertos) * 100 
    varianza = np.var(aciertos)     
    
    print(f"{nombre:<25} | {media:.2f} {'%':<19} | {varianza:<10.6f}")

print("-" * 65)
```
