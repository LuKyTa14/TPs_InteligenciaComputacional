```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import KFold, cross_val_score
from sklearn.ensemble import BaggingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier

# Datos
wine = load_wine()
X, y = wine.data, wine.target

# KFold (K=5)
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Modelos
modelos = {
    "Bagging (50 Árboles)": BaggingClassifier(estimator=DecisionTreeClassifier(random_state=42),n_estimators=50),
    "AdaBoost (50 Tocones)": AdaBoostClassifier(estimator=DecisionTreeClassifier(max_depth=1, random_state=42),n_estimators=50)
}

print("-" * 65)
print(f"{'Ensamble':<25} | {'Media de Acierto':<25} | {'Varianza':<25}")
print("-" * 65)

# Entrenamiento y Evaluacion
for nombre, modelo in modelos.items():
    scores = cross_val_score(modelo, X, y, cv=kf)
    
    media = np.mean(scores) * 100
    varianza = np.var(scores)

    print(f"{nombre:<25} | {media:.2f} {'%':<19} | {varianza:<10.6f}")

print("-" * 65)
```

```python
print("-" * 65)
print(f"{'Ensamble':<25} | {'Media de Acierto':<25} | {'Varianza':<25}")
print("-" * 65)

# Entrenamiento y Evaluacion
for nombre, modelo in modelos.items():
    aciertos = []
    
    # Llamada directa al split
    for train_idx, test_idx in kf.split(X):
        # modelo = clone(modelo)
        modelo.fit(X[train_idx], y[train_idx])
        acierto = modelo.score(X[test_idx], y[test_idx])
        aciertos.append(acierto)

    aciertos = np.array(aciertos) 
    media = np.mean(aciertos) * 100 
    varianza = np.var(aciertos)
    
    print(f"{nombre:<25} | {media:.2f} {'%':<19} | {varianza:<10.6f}")

print("-" * 65)
```
